### AI Foundry の AOAIサービスに対してresponses API(V1)でリクエストに修正
前の応答の response.id を previous_response_id パラメーターに渡すことで、履歴を参照した対話にする。

```python
import os
from openai import OpenAI

API_KEY = os.getenv("AZURE_OPENAI_API_KEY","").strip()
assert API_KEY, "ERROR: Azure OpenAI Key is missing"

RESOURCE_ENDPOINT = os.getenv("AZURE_OPENAI_ENDPOINT_V1","").strip()
assert RESOURCE_ENDPOINT, "ERROR: Azure OpenAI Endpoint is missing"

deployment = os.getenv("AZURE_OPENAI_DEPLOYMENT", "").strip()
assert deployment, "ERROR: Azure OpenAI Deployment name is missing"

client = OpenAI(
    api_key=API_KEY,
    base_url=RESOURCE_ENDPOINT
)
```

##### １回目の応答

```python
# Create your first prompt
first_text_prompt = "日本で一番大きい湖は？"

first_response = client.responses.create(
  model=deployment,
  instructions="You are a helpful assistant.",
  input=first_text_prompt
)

print(first_response.output_text)
print(first_response.id)
```

##### ２回目の応答

```python
second_text_prompt = "２番目は？"

seconde_response = client.responses.create(
    model=deployment,
    previous_response_id=first_response.id,
    input=second_text_prompt
)
print(seconde_response.output_text)
print(seconde_response.id)
```

##### 3回目の応答

```python
third_text_prompt = "それぞれについて名産や盛んなレクリエーションをあげてください？"

third_response = client.responses.create(
    model=deployment,
    previous_response_id=seconde_response.id,
    input=third_text_prompt
    )
print(third_response.output_text)
print(third_response.id)
```
