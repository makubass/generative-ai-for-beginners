```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
## - pre-reqs: install openai and azure-ai-projects packages
##   pip install openai azure-identity azure-ai-projects==1.0.0
## - deploy a gpt-5-nano model

from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient

project_client = AIProjectClient(
    # endpoint="https://YOUR-RESOURCE-NAME.services.ai.azure.com/api/projects/YOUR_PROJECT_NAME",
    endpoint="https://msf-makbs-aifoundry.services.ai.azure.com/api/projects/my-first-project",
    credential=DefaultAzureCredential(),
)

models = project_client.get_openai_client(api_version="2024-10-21")

response = models.chat.completions.create(
    model="gpt-5-nano",
    messages=[
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": "日本で3番目に高い山は？"},
    ],
)

print(response.choices[0].message.content)
```
